# <> Creating Random Tensors

```python
# Creating Random Tensors

# Random tensors are tensors of arbitrary size which contain random numbers.
# They are often used to initialize the weights of neural networks.

# Create a random tensor with a shape of (3, 5)
random_tensor = tf.random.uniform(shape=(3, 5))
display(random_tensor)
```

```python
# Create a random tensor using tf.random.Generator.from_seed
# seed means create random but flavour them with seed
random_tensor_seeded = tf.random.Generator.from_seed(42).uniform(shape=(3,2))
random_tensor_seeded2 = tf.random.Generator.from_seed(42).normal(shape=(3,2))

display(random_tensor_seeded)
display(random_tensor_seeded2)
random_tensor_seeded == random_tensor_seeded2
```

The difference between `tf.random.uniform` and `tf.random.Generator.from_seed`:

*   `tf.random.uniform` generates a new random tensor each time it is called.
*   `tf.random.Generator.from_seed` creates a generator with a fixed seed. When you generate tensors using this generator, you will get the same "random" sequence of numbers each time, as long as the seed is the same. This is useful for reproducibility in experiments.

Both lines use tf.random.Generator.from_seed(42) to ensure that even though the numbers are random, they will be the same "random" numbers each time you run the code because of the fixed seed (42).

The difference comes from .uniform() and .normal():

.uniform(shape=(3,2)): This generates a tensor of shape (3, 2) where each number is picked randomly from a uniform distribution. This means every number within a certain range has an equal chance of being selected. If you were to plot many numbers from a uniform distribution, you'd get a flat-looking shape.
.normal(shape=(3,2)): This generates a tensor of shape (3, 2) where each number is picked randomly from a normal distribution (also known as a Gaussian distribution or bell curve). In this distribution, numbers closer to the average (mean) are more likely to be selected, and numbers further away are less likely. If you were to plot many numbers from a normal distribution, you'd get a bell shape.
Even though both use the same seed, because they use different methods (.uniform() vs .normal()), they generate completely different sets of random numbers, as you saw in the output where random_tensor_seeded == random_tensor_seeded2 resulted in all False.

> numpy array to tensors

*   **Tensors** are the fundamental data structures in TensorFlow. They are similar to NumPy arrays but have key differences:
    *   Tensors can be used on GPUs or TPUs for faster computation.
    *   Tensors are immutable by default (cannot be changed after creation) unless they are `tf.Variable`.
    *   Tensors can be automatically differentiated (useful for training neural networks).

*   **NumPy arrays** are a fundamental package for scientific computing in Python.
    *   NumPy arrays run on the CPU.
    *   NumPy arrays are mutable (can be changed after creation).
    *   NumPy arrays do not have built-in automatic differentiation capabilities.

```python
import numpy as np
numpy_A = np.arange(1,25 , dtype=np.int32)
numpy_A
```

```python
A = tf.constant(numpy_A , shape = (2,3,4))
B = tf.constant(numpy_A)
A , B
```
